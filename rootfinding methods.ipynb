## Secant Method

```python
from datetime import datetime as dt
import math
def f(x):
     return math.cos(x)-x*math.exp(x) 
# a=float(input("the value of a "))
# b=float(input("the value of b "))
# e=float(input(" tolerence "))
a=-2
b=-1
e=0.00001
itime=dt.now()  
i=1
z=1
while z>e:
    c=b-((f(b)*(b-a))/(f(b)-f(a))) 
    z=abs(b-c)
    a=b
    b=c
    print("At",i,'iteration appr root is ',c,"error is",z)   
    i=i+1
print("app root",c)
ftime=dt.now()
print("The cost of computation is ",ftime-itime)
```

## Newton Methods

```python
import sympy as smp
from sympy import *
from datetime import datetime as dt

def f(x1):
    return smp.cos(x1)-x1*smp.exp(x1)
a=29.59
e=0.001
itime=dt.now()  
i=1
z=0.5
while z>e:
    x=symbols('x')
    dx=diff(f(x),x)
    dx_val=smp.lambdify(x,dx)
    b=a-(f(math.radians(a))/dx_val(math.radians(a)))
    z=abs(b-a)
    a=b
    print("At",i,'iteration appr root is ',b,"error is",z)
    i=i+1
print("app root",a)
ftime=dt.now()
print("The cost of computation is ",ftime-itime)
```

## Newton Raphson method with multiplicity 

```python
x0 = 3   # Initial guess  
z = 2   # Tolerance (4 significant digits)  
m = 2
itime=dt.now() 
e=10**(-z)
x=symbols('x')
def f(x):
    return x**2-1
dfdx=diff(f(x),x)
dfdx_val=smp.lambdify(x,dfdx)
if dfdx_val(x0)!=0:
    h=((f(x0))/dfdx_val(x0))
    i=0
    while abs(h)<=e:
        if dfdx_val(x0)!=0:
            h=((f(x0))/(dfdx_val(x0)))
            x1=x0-m*h
            print(f'iteration {i+1}',x1)
            x0=x1
        i+=1
    print(f'The root of {f(x)} is :',x1)
ftime=dt.now()
print("The cost of computation is ",ftime-itime)
```
